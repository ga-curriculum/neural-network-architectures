## Dataset: Stanford Dogs
The Stanford Dogs dataset contains over 20,000 images across 120 dog breeds.

**Download:** [Stanford Dogs Dataset](http://vision.stanford.edu/aditya86/ImageNetDogs/)

```python
# Import Libraries
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
```

### Load and Preprocess the Dataset

```python
# Define transforms for data augmentation and normalization
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Load Stanford Dogs dataset (example placeholders)
train_set = datasets.FakeData(transform=transform)  # Replace with actual dataset
train_loader = DataLoader(train_set, batch_size=16, shuffle=True)

# Visualize sample images
data_iter = iter(train_loader)
images, labels = next(data_iter)
plt.figure(figsize=(8, 8))
for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i].permute(1, 2, 0))
    plt.axis('off')
plt.show()
```

### Load a Pre-trained Model

```python
# Load a pre-trained ResNet model
model = models.resnet18(pretrained=True)

# Freeze the feature extractor
for param in model.parameters():
    param.requires_grad = False

# Replace the classifier for our task
num_classes = 120  # Number of dog breeds
model.fc = nn.Linear(model.fc.in_features, num_classes)
print(model)
```

### Fine-Tune the Model

```python
# Define loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)

# Training loop (placeholder example)
for epoch in range(5):
    for images, labels in train_loader:
        outputs = model(images)
        loss = criterion(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")
```
